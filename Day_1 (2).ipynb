```python
# data Preprocessing
# 1. Cheack the missing values
# 2. Handle the missing values
# 3. Drop the columns
# 4. Check the duplicates
# Remove those duplicates
```

```python
data.isnull().sum()
```

```python
data.drop('Cabin', axis=1, inplace=True)

data['Age'].fillna(data['Age'].median(), inplace=True)

mode_embarked = data['Embarked'].mode()[0]
data['Embarked'].fillna(mode_embarked, inplace=True)

print("Missing values handled. Here's the updated count of missing values:")
```

```python
data.isnull().sum()
```

```python
duplicates_count = data.duplicated().sum()
print(f"Number of duplicate rows found: {duplicates_count}")
```
