```python
# BFS
graph = {
    'A' : ['B', 'C'],
    'B' : ['D', 'E'],
    'C' : ['F', 'G'],
    'D' : [],
    'E' : [],
    'F' : [],
    'G' : []
}
```

```python
from collections import deque
```

```python
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    while queue:
        node = queue.popleft()
        if node not in visited:
            print(node, end=" ")
            visited.add(node)
            queue.extend(graph[node])
```

```python
bfs(graph, 'A')
```

```python
import heapq
```

```python
graph = {
    'S' : [('A', 1), ('B', 4)],
    'A' : [('G', 2)],
    'B' : [('G', 5)],
    'G' : []
}

heuristic = {
    'S' : 3,
    'A' : 2,
    'B' : 4,
    'G' : 0
}
print(graph)
print(heuristic)
```

```python
def astar(start, goal):
    queue = []
    heapq.heappush(queue, (0, start))
    cost = {start: 0}
    while queue:
        current = heapq.heappop(queue)[1]
        print(current)
        if current == goal:
            print("Goal Reached")
            break
        for neighbor, weight in graph[current]:
            new_cost = cost[current] + weight
            if neighbor not in cost or new_cost < cost[neighbor]:
                cost [neighbor] = new_cost
                priority = new_cost + heuristic[neighbor]
                heapq.heappush(queue, (priority, neighbor))
```

```python
astar('S', 'G')
```
