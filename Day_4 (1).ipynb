```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
# Import KMeans
from sklearn.cluster import KMeans
```

```python
# Import matplotlib
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
iris
```

```python
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data
```

```python
data.head(5)
```

```python
# Select features
x = data
```

```python
# Create KMeans model
model = KMeans(n_clusters=3, random_state=42)
```

```python
# Train the model
model.fit(x)
```

```python
# Print cluster centers
print("CLuster Centers\n")
print(model.cluster_centers_)
```

```python
# Predict clusters
clusters = model.predict(x)
```

```python
# Add clusters labels
data['Cluster'] = clusters
```

```python
# Display first five rows
print(data.head(5))
```

```python
# Plot clusters
plt.figure(figsize=(8, 6))
plt.scatter(
    data.iloc[:,2],
    data.iloc[:,3],
    c=data['Cluster'],
    cmap='Accent'  #rainbow, viridis, magam, inferno, cool, hot, plasma, Accent
)
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('K-Means Clustering using Iris Dataset')
plt.show()

# Display number of samples in each cluster
print("\n____________________________________________________________________\n")
print("\nCluster Counts\n")
print(data['Cluster'].value_counts())
```

```python
cluster_counts = data['Cluster'].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(cluster_counts, labels=cluster_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Clusters')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
