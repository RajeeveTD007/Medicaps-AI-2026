```python
# 7. DATA PREPROCESSING

# Check missing values
print("Missing values:")
print(df.isnull().sum())


# Remove duplicate rows
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# Handle missing numerical values
numeric_columns = [
    "area",
    "beds",
    "bathrooms",
    "balconies",
    "area_rate"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())


# Handle missing categorical values
categorical_columns = [
    "house_type",
    "locality",
    "city",
    "furnishing"
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])


# Handle missing target values
df = df.dropna(subset=["rent"])


# Remove invalid values
df = df[
    (df["area"] > 0) &
    (df["beds"] > 0) &
    (df["bathrooms"] > 0) &
    (df["rent"] > 0)
]


# Handle outliers in rent using IQR
Q1 = df["rent"].quantile(0.25)
Q3 = df["rent"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df["rent"] >= lower_limit) &
    (df["rent"] <= upper_limit)
]

print("Shape after handling outliers:", df.shape)


# Encode categorical columns
df = pd.get_dummies(
    df,
    columns=["house_type", "locality", "city", "furnishing"],
    drop_first=True
)

print("Shape after encoding:", df.shape)
```

```python
# 8. DEFINE X & y

X = df.drop("rent", axis=1)

y = df["rent"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# 9. TRAIN / TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# 10. BASELINE MODEL

from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
# 11. TRAIN MODEL

model.fit(X_train, y_train)

print("Model training completed.")
```

```python
# 12. PREDICT

y_pred = model.predict(X_test)

print("Predicted rent:")
print(y_pred[:10])
```

```python
# 13. EVALUATE MODEL

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)


print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# 14. FEATURE ENGINEERING

# Total number of rooms
df["total_rooms"] = df["beds"] + df["bathrooms"]


# Area per bedroom
df["area_per_bed"] = df["area"] / df["beds"]


# Area per bathroom
df["area_per_bathroom"] = df["area"] / df["bathrooms"]

print(df[[
    "area",
    "beds",
    "bathrooms",
    "total_rooms",
    "area_per_bed",
    "area_per_bathroom"
]].head())
```

```python
#Transform Features
df["log_area"] = np.log1p(df["area"])

# Remove irrelevant columns if they exist
columns_to_remove = []

if "id" in df.columns:
    columns_to_remove.append("id")

df = df.drop(columns=columns_to_remove)

df
```

```python
# 15. MODEL IMPROVEMENT

# Separate features and target

X = df.drop("rent", axis=1)

y = df["rent"]
```

```python
lr = LinearRegression()

lr.fit(X_train, y_train)

lr_pred = lr.predict(X_test)
```

```python
# Ridge Regression

from sklearn.linear_model import Ridge

ridge = Ridge(alpha=1.0)

ridge.fit(X_train, y_train)

ridge_pred = ridge.predict(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
```

```python
# 16. MODEL COMPARISON

# Calculate metrics

# Linear Regression
lr_mae = mean_absolute_error(y_test, lr_pred)
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))
lr_r2 = r2_score(y_test, lr_pred)


# Ridge Regression
ridge_mae = mean_absolute_error(y_test, ridge_pred)
ridge_rmse = np.sqrt(mean_squared_error(y_test, ridge_pred))
ridge_r2 = r2_score(y_test, ridge_pred)


# Random Forest
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)
```

```python
# Create comparison table

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Random Forest"
    ],

    "MAE": [
        lr_mae,
        ridge_mae,
        rf_mae
    ],

    "RMSE": [
        lr_rmse,
        ridge_rmse,
        rf_rmse
    ],

    "R2": [
        lr_r2,
        ridge_r2,
        rf_r2
    ]
})

print(results)
```
