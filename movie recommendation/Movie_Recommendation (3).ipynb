```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python

import os
import pickle
import requests
import gradio as gr

# ----------------------------------
# 1. TMDB API configuration
# ----------------------------------

API_KEY = #YOUR APIKEY HERE

# ----------------------------------
# 2. Locate model files
# ----------------------------------

def locate_file(filename):
    possible_paths = [
        f"/content/model/{filename}",
        f"/content/{filename}"
    ]

    for path in possible_paths:
        if os.path.exists(path):
            return path

    raise FileNotFoundError(
        f"{filename} not found. Check your Colab files."
    )


movie_path = locate_file("movie_list.pkl")
similarity_path = locate_file("similarity.pkl")

# ----------------------------------
# 3. Load recommendation model
# ----------------------------------

with open(movie_path, "rb") as f:
    movies = pickle.load(f)

with open(similarity_path, "rb") as f:
    similarity = pickle.load(f)

# Validate data
required_columns = {"movie_id", "title", "tags"}

if not required_columns.issubset(movies.columns):
    raise ValueError("Movie dataset is missing required columns.")

if similarity.shape[0] != len(movies) or similarity.shape[1] != len(movies):
    raise ValueError(
        "Movie list and similarity matrix dimensions do not match."
    )

print("Movie data loaded:", movies.shape)
print("Similarity matrix loaded:", similarity.shape)

# ----------------------------------
# 4. Fetch movie poster from TMDB
# ----------------------------------

def fetch_poster(movie_id):

    url = f"https://api.themoviedb.org/3/movie/{movie_id}"

    params = {
        "api_key": API_KEY,
        "language": "en-US"
    }

    try:
        response = requests.get(
            url,
            params=params,
            timeout=15
        )

        response.raise_for_status()

        data = response.json()

        poster_path = data.get("poster_path")

        if poster_path:
            return (
                "https://image.tmdb.org/t/p/w500/"
                + poster_path
            )

    except requests.exceptions.RequestException as e:
        print("TMDB error:", e)

    return None


# ----------------------------------
# 5. Movie recommendation function
# ----------------------------------

def recommend(movie):

    if not movie:
        return [], "Please select a movie."

    matching_movies = movies[
        movies["title"] == movie
    ]

    if matching_movies.empty:
        return [], "Movie not found."

    index = matching_movies.index[0]

    distances = sorted(
        enumerate(similarity[index]),
        reverse=True,
        key=lambda x: x[1]
    )

    recommended_names = []
    recommended_posters = []

    for i in distances[1:6]:

        movie_id = movies.iloc[i[0]]["movie_id"]
        movie_name = movies.iloc[i[0]]["title"]

        poster = fetch_poster(movie_id)

        recommended_names.append(movie_name)

        if poster:
            recommended_posters.append(
                (poster, movie_name)
            )

    names_text = "### Recommended Movies\n"

    for number, name in enumerate(recommended_names, 1):
        names_text += f"{number}. {name}\n\n"

    return recommended_posters, names_text


# ----------------------------------
# 6. Gradio application interface
# ----------------------------------

movie_list = movies["title"].dropna().tolist()

with gr.Blocks(title="Movie Recommendation System") as demo:

    gr.Markdown(
        """
        # Movie Recommendation System

        Select a movie to discover five similar movies.
        """
    )

    movie_input = gr.Dropdown(
        choices=movie_list,
        label="Search or Select a Movie",
        interactive=True
    )

    recommend_button = gr.Button(
        "Show Recommendations",
        variant="primary"
    )

    recommendation_text = gr.Markdown()

    gallery = gr.Gallery(
        label="Recommended Movie Posters",
        columns=5,
        height=350,
        object_fit="contain"
    )

    recommend_button.click(
        fn=recommend,
        inputs=movie_input,
        outputs=[gallery, recommendation_text]
    )

# ----------------------------------
# 7. Launch application
# ----------------------------------

demo.launch(
    share=True,
    debug=True
)
```
