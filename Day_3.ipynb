```python
#import decision tree classifier
from sklearn.tree import DecisionTreeClassifier
```

```python
#import evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
iris = load_iris()
iris
```

```python
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data
```

```python
# Add target column
data['Species'] = iris.target
iris.target
```

```python
print(data.head())
```

```python
# Separate input and output
x = data.drop('Species', axis=1)
y = data['Species']
data.shape
```

```python
# Split dataset into training and testing data
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42 # Ensures the same split every time code runs.
)
```

```python
# Create Decision Tree model
model = DecisionTreeClassifier()
```

```python
# Train the model
model.fit(x_train, y_train)
```

```python
# Predict test data
y_pred = model.predict(x_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
```

```python
# Display results
print("Accuracy: ", accuracy)

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
ax.figure.colorbar(im, ax=ax)

# We want to show all ticks...
ax.set(xticks=range(cm.shape[1]),
       yticks=range(cm.shape[0]),
       # ... and label them with the respective list entries
       xticklabels=iris.target_names, yticklabels=iris.target_names,
       title='Confusion Matrix',
       ylabel='True label',
       xlabel='Predicted label')

# Rotate the tick labels and set their alignment.
plt.setp(ax.get_xticklabels(), rotation=45, ha="right",
         rotation_mode="anchor")

# Loop over data dimensions and create text annotations.
fmt = 'd'
thresh = cm.max() / 2.
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, format(cm[i, j], fmt),
                ha="center", va="center",
                color="white" if cm[i, j] > thresh else "black")
fig.tight_layout()
plt.show()
```

### Data Visualization using Scatter Plots

Let's visualize the relationships between the features using scatter plots, colored by species, to see how well the different Iris species are separated based on these features.

### 1. Box Plots

Box plots are useful for visualizing the distribution of a continuous variable (like sepal length) across different categories (like species). They show the median, quartiles, and potential outliers.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 10))
for i, feature in enumerate(iris.feature_names):
    plt.subplot(2, 2, i + 1) # Create a 2x2 grid of subplots
    # Plot boxplot for each feature, grouped by 'Species'
    data.boxplot(column=feature, by='Species', ax=plt.gca())
    plt.title(f'Box Plot of {feature.replace(" (cm)", "")} by Species')
    plt.suptitle('') # Suppress the default suptitle generated by pandas boxplot
    plt.xlabel('Species')
    plt.ylabel(feature)

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent overlap
plt.show()
```

### 2. Violin Plots

Violin plots are similar to box plots but also show the probability density of the data at different values, giving a richer display of the distribution shape.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 10))
for i, feature in enumerate(iris.feature_names):
    plt.subplot(2, 2, i + 1)
    # Create a list of data for each species
    plot_data = [data[feature][data['Species'] == j] for j in range(len(iris.target_names))]
    plt.violinplot(plot_data, showmeans=True, showextrema=True, showmedians=True)
    plt.xticks(range(1, len(iris.target_names) + 1), iris.target_names)
    plt.title(f'Violin Plot of {feature.replace(" (cm)", "")} by Species')
    plt.xlabel('Species')
    plt.ylabel(feature)

plt.tight_layout()
plt.show()
```

### 3. Pair Plot (using Seaborn)

A pair plot creates a grid of scatterplots for every pair of variables and histograms for each variable individually. It's excellent for quickly visualizing relationships and distributions across multiple dimensions. For this, `seaborn` is particularly well-suited.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Map the numerical 'Species' back to categorical names for better plotting labels
data['Species_Name'] = data['Species'].map({i: name for i, name in enumerate(iris.target_names)})

sns.pairplot(data, hue='Species_Name', diag_kind='kde') # diag_kind='kde' shows density plots on the diagonal
plt.suptitle('Pair Plot of Iris Dataset by Species', y=1.02) # Adjust suptitle position
plt.show()

# Drop the temporary 'Species_Name' column to keep the DataFrame clean
data.drop(columns=['Species_Name'], inplace=True)
```

### 4. Correlation Heatmap (using Seaborn)

A correlation heatmap visualizes the correlation matrix of the features. It helps in understanding which features are highly correlated with each other, which can be useful for feature selection or understanding multicollinearity.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the correlation matrix, excluding the 'Species' column
correlation_matrix = data[iris.feature_names].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Iris Features')
plt.show()
```
